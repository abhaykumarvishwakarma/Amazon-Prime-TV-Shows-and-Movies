## 3. ***Data Wrangling***

### Data Wrangling Code

Remove Duplicate Values

````python
# Duplicate values before removal

print("Titles Dataset :", titles_df.duplicated().sum())
print("Credits Dataset :", credits_df.duplicated().sum())
````

````python
# Remove duplicates

titles_df = titles_df.drop_duplicates()

credits_df = credits_df.drop_duplicates()
````

````python
# Verify duplicate removal

print("Titles Dataset :", titles_df.duplicated().sum())
print("Credits Dataset :", credits_df.duplicated().sum())
````

Handle Missing Values

````python
# Percentage of missing values of titles dataset

missing_percentage = (titles_df.isnull().sum()/len(titles_df))*100

missing_percentage.sort_values(ascending=False)
````

````python
# Fill missing categorical values

titles_df["age_certification"] = titles_df["age_certification"].fillna("Unknown")
````

````python
# Fill missing numerical values

# Replacing imbd_score, tmdb_score, tmdb_popularity and imdb_votes with median values.

titles_df["imdb_score"] = titles_df["imdb_score"].fillna(titles_df["imdb_score"].median())

titles_df["tmdb_score"] = titles_df["tmdb_score"].fillna(titles_df["tmdb_score"].median())

titles_df["tmdb_popularity"] = titles_df["tmdb_popularity"].fillna(titles_df["tmdb_popularity"].median())

titles_df["imdb_votes"] = titles_df["imdb_votes"].fillna(titles_df["imdb_votes"].median())

# Replacing missing values in the "seasons" column with 0, as it indicates that the title is not a TV show or does not have multiple seasons.
titles_df["seasons"] = titles_df["seasons"].fillna(0)
````

````python
# Fill missing text values

# Replacing missing values in the "description" column with "Description not available" to indicate that the description is not provided for those titles.

titles_df["description"] = titles_df["description"].fillna("Description not available")
````

The 'imdb_id' column was left unchanged because it serves only as an external identifier and is not used in exploratory data analysis or business insights. Therefore, its missing values do not impact the analysis. (I'm not droping this column because if any column has a very high percentage of missing values like more than 70–80% then delete it, and is not useful.)

````python
# Verify missing values of titles dataset after imputation

titles_df.isnull().sum()
````

````python
# Percentage of missing values of credit dataset

missing_percentage_credit = (credits_df.isnull().sum()/len(credits_df))*100

missing_percentage_credit.sort_values(ascending=False)
````

````python
# Replacing missing values in the "character" column with "Unknown" to indicate that the character name is not provided for those credits.

credits_df["character"] = credits_df["character"].fillna("Unknown")
````

### What all manipulations have you done and insights you found?

During the data wrangling phase, both the titles and credits datasets were carefully examined and cleaned to improve data quality before performing Exploratory Data Analysis (EDA). The objective was to ensure that the datasets were accurate, consistent, and suitable for generating meaningful business insights.

1. Dataset Inspection
    * Loaded both titles.csv and credits.csv into Pandas DataFrames.
    * Examined the shape, column names, data types, and summary statistics of each dataset using functions such as head(), info(), describe(), and columns().
    * Identified the different types of variables, including numerical, categorical, and text features.

    Insight:
        Understanding the structure of the datasets helped identify which variables would be useful for analysis and which required preprocessing.

2. Duplicate Value Handling
    * Checked both datasets for duplicate records using the duplicated() function.
    * Removed duplicate rows (if any) using drop_duplicates() to ensure that each observation represented a unique record.

    Why?
        Duplicate records can lead to incorrect counts, biased averages, and misleading visualizations. Removing them ensures that the analysis reflects the actual data.

3. Missing Value Treatment – Titles Dataset

    Several columns in the titles dataset contained missing values. Different strategies were used depending on the purpose of each column.

    * age_certification
        * Missing values were replaced with "Unknown".
        * Since this is a categorical feature, assigning "Unknown" preserves the records without introducing incorrect categories.
    * seasons
        * Missing values were replaced with 0.
        * Movies do not have seasons, so replacing missing values with zero appropriately distinguishes movies from TV shows.
    * imdb_score, imdb_votes, tmdb_score, tmdb_popularity
        * Missing values were replaced using the median of each column.

    - Why Median instead of Mean?
        * These variables may contain extreme values (outliers).
        * The median is more robust to outliers and provides a better estimate of the central tendency.
        * description
            * Missing values were replaced with "Description not available".

    - Why?
        * The description is a text field and may be useful for future NLP or text analysis.
        * Filling it preserves all records while clearly indicating that the description is unavailable.
        * imdb_id
            * Missing values were intentionally left unchanged.
            * imdb_id is only an external identifier and is not used in statistical analysis, visualizations, or business insights.
            * Filling it with artificial values would not add analytical value, and removing the column would unnecessarily discard potentially useful metadata.

4. Missing Value Treatment – Credits Dataset

    The credits dataset had approximately 13% missing values in the character column.

    * Missing values in the character column were replaced with "Unknown".

    - Why?
        * Directors and some crew members do not have associated character names.
        * Filling these missing values preserves the complete dataset while clearly indicating unavailable information.

    Other columns in the dataset did not contain significant missing values and therefore required no additional treatment.

5. Data Type Verification

    - The data types of all variables were verified to ensure consistency.
        * Numerical columns remained as integer or float types.
        * Categorical and text columns remained as object/string types.

    This step confirmed that the datasets were ready for visualization and statistical analysis.

6. Why were the two datasets not merged?

    Although both datasets share a common id column, they were not merged during the data wrangling stage.

    - Reason:
        * The titles dataset contains one record per movie or TV show.
        * The credits dataset contains multiple records for a single title, as each movie or show can have several actors and directors.

    - A direct merge would create duplicate title records because one title would appear once for every associated cast or crew member. This would distort analyses such as:
        * Number of movies by genre
        * Average IMDb rating
        * Runtime distribution
        * Release year trends

Therefore, both datasets were cleaned independently and will remain separate during most of the EDA. They will only be merged later if a specific analysis requires information from both datasets.

This approach maintains data integrity and prevents duplication errors.


- Key Insights from Data Wrangling
    * Both datasets were successfully cleaned and prepared for analysis.
    * Missing values were handled using strategies appropriate to the data type and business context.
    * Duplicate records were removed to avoid biased results.
    * Important text features such as description were preserved rather than discarded.
    * Identifier fields such as imdb_id were retained because they may be useful for future reference, even though they are not used in the current analysis.
    * The titles and credits datasets were intentionally kept separate to avoid many-to-one merge issues and ensure accurate exploratory analysis.
